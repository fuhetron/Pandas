## Merge() with New Dataset

```python
import pandas as pd

students = pd.DataFrame({
    'student_id': [101, 102, 103],
    'name': ['Ali', 'Sara', 'Ahmed']
})

grades = pd.DataFrame({
    'student_id': [101, 102, 104],
    'CGPA': [3.5, 3.8, 3.2]
})
```

## Inner Join using merge()

```python
merged_df = pd.merge(
    students,
    grades,
    on='student_id',
    how = 'inner'
)
print(merged_df)
```

## Left Join

```python
merged_df = pd.merge(
    students,
    grades,
    on='student_id',
    how = 'left'
)
print(merged_df)
```

## Right Join

```python
merged_df = pd.merge(
    students,
    grades,
    on='student_id',
    how = 'right'
)
print(merged_df)
```

## Outer(Full) Join

```python
merged_df = pd.merge(
    students,
    grades,
    on='student_id',
    how = 'outer'
)
print(merged_df)
```

## Concat()

```python
concated_column = pd.concat(
    [students,grades],
    ignore_index=True
)
concated_column
```

# Reshaping Data

## pivot()

#### Changing the structure of data

```python
data = {
    'Student': ['Ali', 'Ali', 'Sara', 'Sara'],
    'Subject': ['Maths', 'English', 'Maths', 'English'],
    'Marks': [85, 90, 78, 88]
}

data = pd.DataFrame(data)
```

```python
data
```

```python
data.pivot(index= 'Student', columns= 'Subject', values= 'Marks')
```

#### The basic functionality is to change the columns to rows to avoid data load. also it doesn't accept duplicate or multiple values , raise ValueError

```python
data.columns.tolist()
```

```python
data.pivot_table(
    index = 'Student',
    columns = 'Subject',
    values= 'Marks',
    aggfunc= 'mean'
)
```

# Pivot table

### Pivot table are pivot but with aggregation function

```python
data_wide = data.pivot(
    index='Student',
    columns='Subject',
    values='Marks'
).reset_index()

data_wide.melt(
    id_vars='Student',
    var_name='Subject',
    value_name='Marks'
)
```
